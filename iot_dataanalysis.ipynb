```python
#import lib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#load iot data
df = pd.read_csv("iot_sensor_data.csv")
df["Timestamp"] = pd.to_datetime(df["Timestamp"])   
print("IOT sensor data \n", df)
```

```python
#Statical Analysis
temp = df["Temperature"].values
mean_temp = np.mean(temp)
std_temp = np.std(temp)
print("\n Mean Temperature : ", mean_temp)
print("Standard Deviation of Temperature : ", std_temp)
```

```python
import matplotlib.pyplot as plt

# Temperature
plt.bar(
    ["Reading 1", "Reading 2", "Reading 3",
     "Reading 4", "Reading 5", "Reading 6"],
    df["Temperature"]
)

plt.xlabel("Reading")
plt.ylabel("Temperature (°C)")
plt.title("Temperature Readings")
plt.xticks(rotation=45)
plt.show()


# Humidity
plt.bar(
    ["Reading 1", "Reading 2", "Reading 3",
     "Reading 4", "Reading 5", "Reading 6"],
    df["Humidity"]
)

plt.xlabel("Reading")
plt.ylabel("Humidity (%)")
plt.title("Humidity Readings")
plt.xticks(rotation=45)
plt.show()


# Pressure
plt.bar(
    ["Reading 1", "Reading 2", "Reading 3",
     "Reading 4", "Reading 5", "Reading 6"],
    df["Pressure"]
)

plt.xlabel("Reading")
plt.ylabel("Pressure (hPa)")
plt.title("Pressure Readings")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(10, 5))
plt.plot(df["Timestamp"], df["Temperature"], label="Temperature", color="tab:red")
plt.plot(df["Timestamp"], df["Humidity"], label="Humidity", color="tab:blue")
plt.title("Temperature and Humidity Trends")
plt.xlabel("Timestamp")
plt.ylabel("Value")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
