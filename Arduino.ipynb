```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("A000066-datasheet.pdf")
pdf_loader = loader.load()
print(pdf_loader[0].page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=200,chunk_overlap=20)
final_docs = text_splitter.split_documents(pdf_loader)
final_docs
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name='all-MiniLM-L6-v2')
```

```python
from langchain_community.vectorstores import FAISS
db = FAISS.from_documents(final_docs, embeddings)
db
```

```python
query = input("Enter your question about the datasheet: ")
documents = db.similarity_search(query)
for i in range(len(documents)):
    print(documents[i].page_content)
```
